#### Text Generation with pre-trained GPT-2

```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer

model_name = "openai-community/gpt2-medium"
model = GPT2LMHeadModel.from_pretrained(model_name)
tokenizer = GPT2Tokenizer.from_pretrained(model_name)

input_text = "What is the capital of Saudi ?"
input_ids = tokenizer.encode(input_text, return_tensors="pt")

output = model.generate(input_ids, max_length=50, num_return_sequences=1, no_repeat_ngram_size=2)
print(tokenizer.decode(output[0], skip_special_tokens=True))
```

```python
tokenizer.decode(output[0])
```

#### Text Classification with BERT

```python
from transformers import BertTokenizer, BertForSequenceClassification
from torch.nn.functional import softmax
import torch

model_name="nlptown/bert-base-multilingual-uncased-sentiment"
tokenizer = BertTokenizer.from_pretrained(model_name)
model = BertForSequenceClassification.from_pretrained(model_name)

text = "I really enjoyed this movie, it was fantastic!"
inputs = tokenizer(text,
                  return_tensors="pt",
                  truncation=True, # Corrected typo from trunction to truncation
                  padding=True)
with torch.no_grad():
  outputs = model(**inputs)
  probabilities = softmax(outputs.logits, dim=-1)
  predicted_class = torch.argmax(probabilities, dim=-1).item()

print(f"Predicted sentiment: {predicted_class}")
```

### Tokenizers

Tokenizers transform textual data into a format the model understands.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = "Hugging Face is leading in the NLP Hugging space. And this bootcamp is amazing."
# Convert text to token ids
input_ids = tokenizer.encode(text, add_special_tokens=True)
print(f"Token IDs: {input_ids}")
```

```python
# Convert token ids back to text
decoded_text = tokenizer.decode(input_ids, skip_special_tokens=False)
print(f"Decoded Text: {decoded_text}")
```
